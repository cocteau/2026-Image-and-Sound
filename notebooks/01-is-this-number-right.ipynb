## Get the data

*This cell is plumbing — you don't need to read it closely.* It downloads the FEC **committee master** file and the FEC's own list of column names.

```python
import io, zipfile, requests, pathlib

data = pathlib.Path("../data"); data.mkdir(exist_ok=True)
zip_path = data / "cm26.zip"
if not zip_path.exists():
    zip_path.write_bytes(requests.get("https://www.fec.gov/files/bulk-downloads/2026/cm26.zip", timeout=60).content)

header = requests.get("https://www.fec.gov/files/bulk-downloads/data_dictionaries/cm_header_file.csv", timeout=60).text
columns = header.strip().split(",")

with zipfile.ZipFile(zip_path) as z:
    with z.open(z.namelist()[0]) as f:
        cm = pd.read_csv(f, sep="|", header=None, names=columns, dtype=str)

cm.head()
```

### Check the plumbing
Even plumbing gets a check. The FEC says this file has **15 columns** and **no header row**. Open the
[committee master file description](https://www.fec.gov/campaign-finance-data/committee-master-file-description/)
and compare the column names above with what the FEC says each one means.

```python
checks.same_number(len(cm.columns), 15, "column count matches FEC description")
```
