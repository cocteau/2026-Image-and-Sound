### Setup
`check` compares what you claimed with what the data says and prints PASS or FAIL. For numbers, `tolerance` says how far off is still OK (for example, a claim that rounds to "about 5,076 g").

```python
import pandas as pd

def check(what, data_says, claimed, tolerance=0):
    if isinstance(claimed, (int, float)):
        ok = abs(data_says - claimed) <= tolerance
    else:
        ok = data_says == claimed
    print("PASS" if ok else "FAIL", "|", what, "| claimed:", claimed, "| data says:", data_says)

penguins = pd.read_csv("../data/penguins.csv")
```

## Claim 1

**Claim:** Among the 342 penguins with a recorded body mass, Gentoo penguins are the heaviest on average, at about 5,076 g, compared with about 3,700 g for Adelie and Chinstrap.

**From:** `01-first-look.ipynb`, worked example.

```python
has_mass = penguins[penguins["body_mass_g"].notna()]
mean_mass = has_mass.pivot_table(index="species", values="body_mass_g", aggfunc="mean")["body_mass_g"]

check("penguins with a recorded mass", len(has_mass), 342)
check("heaviest species on average", mean_mass.idxmax(), "Gentoo")
check("Gentoo mean mass (g)", mean_mass["Gentoo"], 5076, tolerance=1)
check("Adelie mean mass (g)", mean_mass["Adelie"], 3700, tolerance=50)
check("Chinstrap mean mass (g)", mean_mass["Chinstrap"], 3700, tolerance=50)
```
